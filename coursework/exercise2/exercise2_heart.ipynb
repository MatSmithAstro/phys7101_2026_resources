```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import signal, stats, integrate
from scipy.interpolate import CubicSpline

# Data: from a local datasets/ folder if you have the repository, otherwise straight from GitHub (e.g. in Colab).
DATA_URL = "https://raw.githubusercontent.com/MatSmithAstro/phys7101_2026_resources/main/coursework/exercise2/datasets/"
def data_path(filename):
    local = os.path.join("datasets", filename)
    return local if os.path.exists(local) else DATA_URL + filename

beats = pd.read_csv(data_path("fantasia_beats.csv"))        # subject, t_s: time of every heartbeat (s)
resp = pd.read_csv(data_path("fantasia_resp.csv"))          # subject, t_s, resp: breathing at 4 samples per second
people = pd.read_csv(data_path("fantasia_subjects.csv"))    # subject, group (young or elderly), age, sex
FS = 4                                                      # breathing sampling rate (Hz)

def beat_times(subject):
    """Times (s) of every heartbeat for one person."""
    return beats.loc[beats.subject == subject, "t_s"].to_numpy()

def breathing(subject):
    """(t, resp) for one person: time (s) and breathing signal (arbitrary units). A few values may be NaN."""
    d = resp.loc[resp.subject == subject]
    return d["t_s"].to_numpy(), d["resp"].to_numpy()

print(f"{people.shape[0]} people: {(people.group == 'young').sum()} young, {(people.group == 'elderly').sum()} elderly; "
      f"{len(beats)} beats in total")
```

```python
# Optional, for extensions: the raw recordings (ECG and breathing at 250 samples per second; 333 for f2y02),
# straight from PhysioNet. Needs an internet connection; installs the wfdb package the first time.
RUN_RAW = False                 # set to True to try it

if RUN_RAW:
    import subprocess, sys
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "wfdb"], check=True)
    import wfdb

    BEAT = set("NLRBAaJSVrFejnE/fQ?")          # annotation symbols that mark a heartbeat

    def raw_recording(sub, t0=1800, t1=3600, pn_dir="fantasia"):
        """Raw signals for one person from t0 to t1 seconds into the recording (minutes 30-60, the same segment as
        the CSV files). Returns time (s, from t0), a dict of signals (RESP, ECG, and BP for some people),
        the annotated beat times (s, from t0) and the sampling rate."""
        fs = wfdb.rdheader(sub, pn_dir=pn_dir).fs
        a, b = int(t0 * fs), int(t1 * fs)
        rec = wfdb.rdrecord(sub, pn_dir=pn_dir, sampfrom=a, sampto=b)
        ann = wfdb.rdann(sub, "ecg", pn_dir=pn_dir, sampfrom=a, sampto=b)
        keep = np.array([s in BEAT for s in ann.symbol])
        signals = {name: rec.p_signal[:, k] for k, name in enumerate(rec.sig_name)}
        return np.arange(rec.sig_len) / fs, signals, ann.sample[keep] / fs - t0, fs

    t_raw, raw, beats_raw, fs_raw = raw_recording(SUB)
    print(f"{SUB}: {fs_raw} samples per second; signals {list(raw)}; {len(beats_raw)} annotated beats")
    print("same beats as fantasia_beats.csv:", len(beats_raw) == len(beat_times(SUB)),
          f"(largest time difference {1000 * np.max(np.abs(beats_raw - beat_times(SUB))):.1f} ms)" if len(beats_raw) == len(beat_times(SUB)) else "")
    w = (t_raw > 600) & (t_raw < 610)
    fig, ax = plt.subplots(2, 1, figsize=(10, 4), sharex=True)
    ax[0].plot(t_raw[w], raw["ECG"][w], lw=0.6, color="k")
    for tb in beats_raw[(beats_raw > 600) & (beats_raw < 610)]: ax[0].axvline(tb, color="#B5121B", lw=0.6, ls=":")
    ax[1].plot(t_raw[w], raw["RESP"][w], color="#1F5FA8")
    ax[0].set(ylabel="ECG (mV)", title=f"{SUB}: 10 s of the raw recording, annotated beats dotted"); ax[1].set(xlabel="Time from minute 30 (s)", ylabel="breathing")
    plt.tight_layout(); plt.show()
```
