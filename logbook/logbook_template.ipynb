* **Date, session and notebook** (with a link).

* **Prediction:** what you expect, written *before* you run it.
* **What happened:** the result, with the numbers.
* **Problems and fixes:** what broke and how you fixed it.
* **Decisions:** choices you made, and why.
* **Questions:** things to ask about or come back to.
* **Figures:** any plot worth keeping. Right-click the plot in your workshop notebook → Copy image, then paste it into a text cell here, with a one-line caption.
