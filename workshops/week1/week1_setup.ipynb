## Why notebooks, and why the cloud?

This module uses two tools that are now standard across scientific research and industry data work:

* **Interactive notebooks (Jupyter).** Code, results, plots and written explanation live in one document, so an analysis can be re-run, checked and read by someone else, including your future self.
* **Cloud computing (Google Colab).** A free, browser-based place to run Jupyter notebooks, with key Python libraries already installed. Nothing needs to be installed, and the same environment is available to everyone.

Everything in this module is written to work in Google Colab, but it will all run equally well in your own Jupyter installation. If you prefer to use another system (e.g. Spyder) then you are welcome to. There is no one right answer to how to code successfully.

## <span style="color:#B5121B">3. Cells, kernels and execution order</span>

A notebook is made of **cells**. A cell holds either text (Markdown) or code. Code cells are run by the **kernel**, the Python process sitting behind the notebook. Run a cell with the ▶ button, or **Shift + Enter** (run and move to the next cell).

```python
print('Hello, PHYS7101')
```

The kernel remembers everything that has been run, **in the order you ran it, not the order the cells appear on the page**. This is the most common cause of "it worked a minute ago". Try it: run the next two cells, then run the second one *again* a couple of times.

```python
counter = 0
```

```python
counter = counter + 1
print(counter)
```

The same cell gave a different answer each time, because it relies on the (hidden) result  from earlier runs. The number in `[ ]` next to each cell tells you the order things actually ran in.

To fix this, restart and run from the top:

* **Runtime → Restart session** forgets everything (in local Jupyter: **Kernel → Restart**).
* **Runtime → Restart session and run all** is the way to test that your notebook works from end-to-end, top to bottom. Do this before you submit anything.
* **Edit → Clear all outputs** removes printed results and plots, which is useful before sharing a notebook.

### Things to know about Colab sessions

* A Colab session is a temporary virtual machine. It disconnects if you're idle for a while, and runs for at most 12 hours even if you're active.
* When it disconnects, **variables and any files you uploaded or created in the session are lost**. Your notebook itself is safe *if* you saved a copy to Drive.
* After a disconnect, reconnect and use **Runtime → Run all** (or run from the top down to where you were).

## <span style="color:#B5121B">4. How to load the data</span>

Datasets for each week are in the corresponding `datasets/` folder in the repository. If you open only the notebook in Colab, it can't see that folder. To simplify things, every notebook that uses data loads it through this small helper:

* if a local `datasets/` folder is there (you cloned or downloaded the repository), it reads the file from disk;
* otherwise (Colab), it reads the same file directly from GitHub.

Either way you get exactly the same data. Run the next cell: if it prints the first few rows of a table, your setup is working.

You do not have to use this machinery. 

```python
import os
import pandas as pd

DATA_URL = "https://raw.githubusercontent.com/MatSmithAstro/phys7101_2026_resources/main/workshops/week1/datasets/"

def data_path(filename):
    """Local datasets/ folder if present, otherwise the copy on GitHub."""
    local = os.path.join("datasets", filename)
    return local if os.path.exists(local) else DATA_URL + filename

tide = pd.read_csv(data_path("lancaster_quay_tidal.csv"))
print("Loaded from:", data_path("lancaster_quay_tidal.csv"))
tide.head()
```

## <span style="color:#B5121B">5. Writing in Markdown</span>

Text cells use **Markdown**, a very light formatting language. This is commonly used to introduce and motivate what you're trying to achieve, and will most likely be used to write your interpretive statement that goes with each assessment. **Double-click this cell** to see how it's written; Shift + Enter turns it back into formatted text.

| You write | You get |
|---|---|
| `**bold**` or `__bold__` | **bold** |
| `*italic*` or `_italic_` | *italic* |
| `` `code` `` | `code` |
| `# Title`, `## Section`, `### Subsection` | headings of decreasing size |
| `* item` or `- item` at the start of a line | a bulleted list (indent for sub-items) |
| `1. item` | a numbered list |
| `$E = mc^2$` | inline maths: $E = mc^2$ |
| `[link text](https://www.lancaster.ac.uk)` | [link text](https://www.lancaster.ac.uk) |
| `***` on its own line | a horizontal rule |

Equations on a line of their own go between double dollar signs: `$$\int_0^\infty e^{-x}\,dx = 1$$` gives

$$\int_0^\infty e^{-x}\,dx = 1$$

A few things that catch people out:

* **Paragraphs need a blank line** between them; a single line break is ignored.
* Tables are built with `|` and a `|---|---|` row under the header, as in the one above.
* Images can be added with `![description](url)`, or by dragging a file into a text cell in Colab.

For more, see GitHub's [Markdown guide](https://docs.github.com/en/get-started/writing-on-github/getting-started-with-writing-and-formatting-on-github/basic-writing-and-formatting-syntax).

## <span style="color:#B5121B">7. Good habits</span>

* **Restart and run all before you trust a result, and always before you submit.** If it doesn't run top to bottom in a fresh session, it doesn't work (see Section 3).
* **Put a short Markdown cell before a block of code** saying what you're about to do, and **one after** saying what the result means. This module focuses on why you chose a particular option, and what you learnt from it. 
* **Note the *why*, not the *what*.** `# convert to seconds so units match the decay constant` is useful; `# multiply by 60` isn't.
* **Use names you'd understand a year from now**, and where there's an equation, use the same symbols as the notes (`lam` for $\lambda$, `half_life`, ...).
* **Do every calculation in code**, not partly in your head or on paper, so the whole chain can be checked and re-run.
* **Test on a small piece first** (a few rows, a short time range) before running on everything.
* **Google is your friend.** Searching for an error message is a normal part of programming. For most tasks, many people will have faced similar problems. The documentation for the key packages [NumPy](https://numpy.org/doc/stable/), [SciPy](https://docs.scipy.org/doc/scipy/), [pandas](https://pandas.pydata.org/docs/) and [Matplotlib](https://matplotlib.org/stable/) is excellent, but not always easy to follow.
